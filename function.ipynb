```python
x=30# global variable
def show():
    x=20
    print("local variable:" , x)
show()
print("global variable:" , x)
```

```python
x=30 #global variable
def show():
    global x
    print("global variable:",x)
show()
print("global variable:", x)
def display():
     global x
     print("global variable:", x)
display()
```

```python
balance=3000 #global  variable
def show():
    global balance
    deposit=1000
    print("deposit balance:" , deposit)

    balance=balance+deposit #4000
 
show()
print("current balance after deposit: ", balance)
```

```python
balance=3000
def show():
    deposit=1000
    print("deposit balance:", deposit)
    print("initial balance was:", globals() ["balance"])
show()
print("current balance after deposit:", balance)
```

```python
def add(a,b):
    return a+b
def add(a,b,c):
    return a+b+c
def add(a,b,c,d):
    return a+b+c+d
def add(a,b,c,d,e):
    return a+b+c+d+e

add(10,20) #2
add(10,20,30) #3
add(10,20,30,40) #4
add(10,20,30,40,50) #5
```

```python
#args
def add(*args): #tuple
    total=0 #30 
    for i in args:
        total=total+i #150
    print("Sum:", total)

add(10,20)
add(10,20,30)
add(10,20,30,40)
add(10,20,30,40,50)
add(10,20,30,40,50,60)
```

```python
#args
def add(*args): #tuple
    total=0 #30
    for i in args:
        total=total+i #450
    print("sum:", total)

add(10,20)
add(40,50,60)
add(20,30,40,50)
add(20,30,40,50,60)
```

```python
def Employeedetails(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)

    print("=======================")
Employeedetails(Name="Mohit",Role="Developer",Department="IT")
Employeedetails(Name="Mohit",Department="IT")
Employeedetails(Name="Mohit")
```

```python
def shopping_bill(*args):
    total = 0

    for price in args:
        total = total + price

    print("Prices:", args)
    print("Total Bill: ₹", total)

shopping_bill(100, 200, 300)
```
