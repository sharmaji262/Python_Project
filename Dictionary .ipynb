```python
#containing student marks,use loop to print whether each student has passed or fail
students={
    'rahul':75,
    'amit':32,
    'priya':88,
    'neha':39,
    'rohit':55}
for name, marks in students.items():
    if (marks>=40):
        print(name,'pass')
    else:
        print(name,'fail')
        
```

```python
students={
    'rahul':85,
    'amit':35,
    'priya':72,
    'neha':48
}
for name, marks in students.items():
    if(marks>=80):
        print(name,'excellent')
    elif(marks>=60):
        print(name,'good')
    elif(marks>=40):
        print(name,'average')
    elif(marks<40):
        print(name,'fail')
```

```python
products={
    'laptop':10,
    'mouse':3,
    'keyboard':0,
    'monitor':7,
    'printer':2
}
for product,stock in products.items():
    if(stock>5):
        if(stock<=5):
            print(product,"low stock")
        elif(stock>5):
            print(product,"available")
    else:
        print(product,"out of stock")
```

```python
employees={
    'rahul':{'salary':60000,'department':'IT'},
    'amit':{'salary':40000,'department':'HR'},
    'sudesh':{'salary':70000,'department':'IT'},
    'priya':{'salary':45000,'department':'finance'}
}
for name,data in employees.items():
    department=data['department']
    salary=data['salary']
    if(salary>=50000):
        if(department=="IT"):
            print(name,'senior IT employee')
        elif(department=="HR"):
            print(name,'senior HR employee')
        else:
            print(name,'senior employee')
    else:
       print(name,'junior employee')
        
```

```python
sales=[
    ["aman",12000,15000,18000],
    ["abhi",22000,11000,21000],
    ["aadi",12000,15000,18900],
]
highest_sale=0
top_salesperson=""


for person in sales:
    name=person[0]
    total=sum(person[1:])

    print("name:",name)
    print("total_sale:",total)
    print("--------------------")

    
    if total>highest_sale:
        highest_sale=total
        top_salesperson=name


print("top_salesperson:",top_salesperson)
print("highest_sale:",highest_sale)
```
